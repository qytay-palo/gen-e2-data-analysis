# PS-001 Workforce Extraction

This notebook imports the PS-001 extraction script, downloads the four SharePoint workforce CSV files into the raw landing zone, and displays the landed file metadata.

```python
import importlib.util
import os
from pathlib import Path

import polars as pl

default_project_root = next(
    path for path in [Path.cwd(), *Path.cwd().parents]
    if (path / '.env').exists() and (path / 'shared').exists()
)
project_root = Path(os.getenv('GEN_E2_PROJECT_ROOT', default_project_root)).expanduser().resolve()
script_path = Path(
    os.getenv(
        'PS001_EXTRACTION_SCRIPT_PATH',
        project_root / 'artifacts' / 'ps-001-workforce-data-foundation' / 'scripts' / 'extract_workforce_data.py',
    )
).expanduser().resolve()
spec = importlib.util.spec_from_file_location('extract_workforce_data', script_path)
module = importlib.util.module_from_spec(spec)
assert spec is not None and spec.loader is not None
spec.loader.exec_module(module)
```

```python
results = module.extract_workforce_files()
summary = pl.DataFrame(
    [
        {
            'file': filename,
            'row_count': details['row_count'],
            'downloaded_at': details['downloaded_at'],
            'path': details['path'],
        }
        for filename, details in sorted(results.items())
    ]
)
summary
```
